## A note on words used in this project

- **District:** a patch of map that elects one representative to the U.S. House.
- **Redistricting:** redrawing district boundaries, done every ten years after the Census.
- **Gerrymandering:** drawing those boundaries to favor one party.
- **Competitive district:** here, a district decided by 5 percentage points or less. That threshold is our choice, and we say so wherever it matters.
